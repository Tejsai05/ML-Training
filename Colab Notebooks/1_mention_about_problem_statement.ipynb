PROBLEM STATEMENT

Access to clean and safe drinking water is essential for human health. However, water contamination remains a significant concern, affecting millions of people worldwide. Various physical and chemical factors determine water potability, including pH levels, hardness, solids, chloramines, sulfates, conductivity, organic carbon, trihalomethanes, and turbidity.This project aims to develop a machine learning model that can accurately predict whether water is potable (safe to drink) or not based on given water quality parameters. Using a dataset of 81,000 water samples, the model will analyze trends, identify critical factors affecting water quality, and provide insights into water safety standards.

COLUMNS

dataset contains 10 columns, each representing a different water quality parameter. Here's what they mean:

pH : Measures how acidic or basic the water is. A pH between 6.5 and 8.5 is generally considered safe for drinking.

Hardness : Refers to the concentration of calcium and magnesium in water. High hardness can cause scaling in pipes but is not harmful to health.

Solids (Total Dissolved Solids - TDS) : The amount of dissolved substances (salts, minerals, metals) in water, measured in ppm (parts per million). Safe drinking water typically has TDS < 500 ppm.

Chloramines : A disinfectant used in water treatment. High levels can be harmful but are used to control microbial growth.

Sulfates : Found naturally in water; excessive levels can cause a laxative effect and impact taste.

Conductivity : Indicates water’s ability to conduct electricity, which depends on the number of dissolved ions. Higher conductivity means more dissolved salts.

Organic Carbon : Measures the amount of organic matter in the water, which can come from decaying plants or industrial waste. High levels indicate pollution.

Trihalomethanes (THMs) :A byproduct of chlorine disinfection, high levels can be harmful to health.

Turbidity : Refers to water clarity; high turbidity means more suspended particles, which can indicate contamination.

Potability : The target variable (0 = Not Safe to Drink, 1 = Safe to Drink).
