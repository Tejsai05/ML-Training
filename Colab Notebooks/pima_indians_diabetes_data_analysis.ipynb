RANDOM FOREST ACCURACY CALCULATION THROUGH DIRECT METHOD

```python
import pandas as pd
import numpy as np
from pandas import read_csv
from sklearn.model_selection import KFold #used to segregate data points
from sklearn.model_selection import cross_val_score #used for checking validation score i.e accuracy score for each tree
from sklearn.ensemble import RandomForestClassifier #used for randomforest model
```

```python
df1=pd.read_csv('pima-indians-diabetes.data.csv')
df1
```

```python
from os import name
data_file='/content/pima-indians-diabetes.data.csv'
names=['preg','plas','pres','skin','test','mass','pedi','age','class']
df=pd.read_csv(data_file,names=names)
```

```python
data_file
```

```python
array=df.values
```

```python

array
```

```python
X=array[:,0:8]
Y=array[:,8]
```

```python
X
```

```python
Y
```

```python
## parameters to model i.e the no.of trees to be created is depends on us
seed=7
num_trees=100
k_fold=KFold(n_splits=10,random_state=seed,shuffle=True)
model=RandomForestClassifier(n_estimators=num_trees,max_features=3)
results=cross_val_score(model,X,Y,cv=k_fold) #fold is referred as tress in a decision tree
```

```python
results
```

```python
print(results.mean())
```
