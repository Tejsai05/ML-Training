```python
for i in range(6):
  for j in range(5):
    print("* ",end="")
  print()
```

```python
#i represents no.of rows
#j reprsents no.of cols
for i in range(5):
  for j in range(i):
    print("* ",end="")
  print()
```

```python
for i in range(5):
  for j in range(i+1):
    print("* ",end="")
  print()
```

```python
for i in range(5):
  for j in range(5-i):
    print("* ",end="")
  print()
```

```python
for i in range(5):
  for j in range((i+1)*2):
    print("* ",end="")
  print()
```
